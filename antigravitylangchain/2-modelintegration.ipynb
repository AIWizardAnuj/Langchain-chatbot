Streaming and Batching

STREAMING

```python
model.invoke("write 200 words paragraph on artificial intelligence")
```

```python
for chunk in model.stream("write 200 words paragraph on artificial intelligence"):
   print(chunk.text, end="",flush=True)
```

BATCHING

```python
responses = model.batch(["why airplane flly?", "who is prime minister of India?","what is artificial intelligence"
])

for response in responses:
    print(response)
```

```python
model.batch(
    ["Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)
```
